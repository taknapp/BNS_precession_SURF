```python
import numpy as np 
import matplotlib.pyplot as plt
import multiprocessing, pathos.multiprocessing

from tqdm import tqdm #usefull to check progress
CPUS=10
#CPUS=int(multiprocessing.cpu_count()) use all the avaible cpu in computer/cluster
parmap = pathos.multiprocessing.ProcessingPool(CPUS).imap
```

```python
def func(x,y):
    a=x*y+4
    b=1
    return a,b

x_arr=np.linspace(0,100,100)
y_arr=np.linspace(0,3,100)
#x and y must be of the same lenght
res =np.array(list(tqdm(parmap(func,x_arr,y_arr))))
```

```python
a_arr=res.T[0]
b_arr=res.T[1]
#random plotting jsut to show that works
plt.figure()
plt.plot(x_arr, a_arr)
plt.plot(y_arr, a_arr)

plt.figure()
plt.plot(x_arr, b_arr)
plt.plot(y_arr, b_arr)
```
